```python
#problem 3
scores = [45, 67, 82, 59]

def pass_fail(score):
    if score >= 60:
        return "Pass"
    else:
        return "Fail"

for score in scores:
    result = pass_fail(score)
    print(f"Input: {score} -> Output: {result}")
```

```python
#problem 4
# AIML 201 Lab

sizes = [500, 800, 1000, 1200]
prices = [50, 80, 100, 130]

for i in range(len(sizes)):
    print(
        "Input:", sizes[i], "square meter -> Output:",
        prices[i], "thousand dollars"
    )

is_increasing = True

for i in range(1, len(prices)):
    if prices[i] < prices[i - 1]:
        is_increasing = False

if is_increasing:
    print("Positive trend")
else:
    print("Negative trend")
```

```python
#problem 5

study_hours = [2, 4, 6, 8]
scores = [55, 65, 75, 85]

total_increase = 0

for i in range(1, len(scores)):
    total_increase += scores[i] - scores[i - 1]

average_increase = total_increase / (len(scores) - 1)

predicted_score = scores[-1] + average_increase

print("Average increase per 2 hours:", average_increase)
print("Predicted score for 10 hours:", predicted_score)
```

```python
#problem 6
scores = [35, 58, 62, 74, 89, 45]

for score in scores:
    if score < 50:
        category = "Low"
    elif 50 <= score <= 70:
        category = "Medium"
    else:
        category = "High"

    print(f"Score: {score} -> Category: {category}")
```

```python
#problem 7
ages = [18, 20, 22, 24, 26]
heights = [160, 165, 170, 175, 180]

for age, height in zip(ages, heights):
    print(f"Age: {age} years -> Height: {height} cm")

average_age = sum(ages) / len(ages)
print("Average age:", average_age)

average_height = sum(heights) / len(heights)
print("Average height:", average_height)

if heights == sorted(heights):
    print("Height appears to increase with age.")
else:
    print("Height does not show a clear increasing trend with age.")
```
